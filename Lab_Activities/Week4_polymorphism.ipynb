## Part 0 — Recap: one class on its own

Here is a single `Shape` class, like the ones from Weeks 2–3. Run it — then notice the problem: if we also want a `Circle` and a `Square`, we'd copy most of this code. Inheritance fixes that.

```python
class Shape:
    def __init__(self, name):
        self.name = name

    def describe(self):
        print(f'I am a {self.name}.')

s = Shape('shape')
s.describe()          # I am a shape.
```

## Part 1 — Inheritance: build a child class  ·  ① Inheritance

> **From the slides — Concept 1 (Inheritance):** *define a new class as a modified version of an existing one.* The trick is putting the parent's name **in parentheses**.

`Circle` inherits everything from `Shape` — including `describe()` — without rewriting it. Run this:

```python
class Circle(Shape):        # Circle inherits from Shape
    pass                    # adds nothing yet — but already has describe()

c = Circle('circle')
c.describe()                # I am a circle.   <- inherited method!
```

> Notice `Circle` has a working `describe()` method that we **never wrote** — it came from `Shape`. That is inheritance.

**Your turn.** Make a class `Square` that inherits from `Shape` (just like `Circle`). Create a `Square('square')` and call `describe()` on it.

```python
class Square(Shape):        # Square inherits from Shape
    def describe(self):      # overrides the inherited method
        print(f'I am a {self.name}, and I have 4 sides.')

s = Square('square')
s.describe()
```

## Part 3 — Give the child its own data with super()  ·  ③ super()

> **From the slides — Concept 3 (super()):** *let the child do everything the parent does first, then add its own part.*

A circle needs a **radius** as well as a name. In `__init__`, we call `super().__init__(...)` to run the parent's setup (which stores the name), then add the radius. Run this:

```python
class Circle(Shape):
    def __init__(self, radius):
        super().__init__('circle')   # run Shape.__init__ -> sets self.name
        self.radius = radius         # then add the child's own attribute

c = Circle(5)
c.describe()          # I am a circle.   (from the parent)
print(c.radius)       # 5               (from the child)
```

> Without `super().__init__('circle')`, the object would have **no** `name`, and `describe()` would fail. `super()` reuses the parent's work instead of copying it.

**Your turn.** Write a `Rectangle(Shape)` whose `__init__` takes `width` and `height`. Call `super().__init__('rectangle')`, then store `self.width` and `self.height`. Create one and call `describe()`.

```python
class Rectangle(Shape):
    def __init__(self, width, height):
        super().__init__('rectangle')
        self.width = width
        self.height = height

r = Rectangle(10, 5)
r.describe()
```

## Part 4 — Overriding: change a method in the child  ·  ③ Override

> **From the slides — Concept 3 (Overriding):** *a child can define a method with the same name to replace the parent's version.*

We give `Shape` an `area()` method that returns `0`, then **override** it in `Circle` with the real formula. Run this:

```python
class Shape:
    def __init__(self, name):
        self.name = name
    def area(self):
        return 0                     # generic default

class Circle(Shape):
    def __init__(self, radius):
        super().__init__('circle')
        self.radius = radius
    def area(self):                  # OVERRIDE: same name, new behaviour
        return 3.14159 * self.radius ** 2

print(Shape('x').area())    # 0        (parent's version)
print(Circle(5).area())     # 78.53975 (child's override)
```

> Same method name, `area()`, but a `Circle` object runs the **child's** version. That is overriding.

**Your turn.** Add `Rectangle(Shape)` again, this time with an `area()` that returns `self.width * self.height`. Test it: a `Rectangle(3, 4)` should give `12`.

```python
class Rectangle(Shape):
    def __init__(self, width, height):
        super().__init__('rectangle')
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

r = Rectangle(3, 4)
print(r.area())  
```

## Part 5 — Polymorphism: one loop, many types  ·  ④ Polymorphism

> **From the slides — Concept 4 (Polymorphism):** *different types share a method name, so one loop works on all of them.*

This is the payoff. Because **every** shape has an `area()` method, we can put different shapes in one list and loop over them — each object runs its own `area()`. First, here is the full hierarchy (the same one from the slides):

```python
class Shape:
    def __init__(self, name):
        self.name = name
    def area(self):
        return 0

class Circle(Shape):
    def __init__(self, radius):
        super().__init__('circle')
        self.radius = radius
    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        super().__init__('rectangle')
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height
```

Now the polymorphic loop — **one line** handles every shape type:

```python
shapes = [Circle(2), Rectangle(3, 4), Circle(1)]

for shape in shapes:
    print(f'{shape.name}: area = {shape.area():.2f}')

# circle: area = 12.57
# rectangle: area = 12.00
# circle: area = 3.14
```

> The loop never checks *which* shape it has. It just trusts every shape has `area()`. Add a new shape type tomorrow and this loop works unchanged — **that** is why polymorphism matters.

**Your turn.** Add a `Triangle(Shape)` class (`__init__` takes `base` and `height`; `area()` returns `0.5 * base * height`). Then add a triangle to the `shapes` list and re-run the loop — it should print the triangle's area with no other changes.

```python
class Triangle(Shape):
    def __init__(self, base, height):
        super().__init__('triangle')
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

shapes = [Circle(2), Rectangle(3, 4), Circle(1), Triangle(5, 4)]

for shape in shapes:
    print(f'{shape.name}: area = {shape.area():.2f}')
```
